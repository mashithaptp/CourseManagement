```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
files_to_load=files.upload()
```

```python
df=pd.read_csv('ABC Company.xlsx - Sheet1.csv')
df.info()
df.isnull().sum()
```

```python
df['Height'] = np.random.randint(150, 181,size=len(df))
df['Height'].value_counts()

```

```python
team_dist = df['Team'].value_counts()
team_perc = df['Team'].value_counts(normalize=True) * 100
plt.figure(figsize=(10,6))
team_dist.plot(kind='bar', color='green')      #bar chart plot
plt.title('Employee Distribution by Team')
plt.ylabel('Number of Employees')
plt.xlabel('Team')
plt.show()
```

Team Dynamics: New Orleans Pelicans team has the highest employee count, suggesting it is the largest department.

```python
position_counts = df['Position'].value_counts()
plt.figure(figsize=(8,8))
plt.pie(position_counts, labels=position_counts.index, autopct='%1.1f%%', startangle=140)          #Pie Chart
plt.title('Employee Distribution by Position')
plt.show()
```

Position Distribution: The company has a high concentration of SG

```python
plt.figure(figsize=(10,6))
sns.histplot(df['Age'], bins=10, kde=True, color='green')
plt.title('Age Distribution of Employees')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

More number of employees are around 20-30 age group

```python
team_salary = df.groupby('Team')['Salary'].sum().sort_values(ascending=False)
pos_salary = df.groupby('Position')['Salary'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,6))
team_salary.plot(kind='bar', color='orange')
plt.title('Total Salary Expenditure by Team')
plt.ylabel('Total Salary')
plt.show()
```

The highest salary expenditure is in the Cleveland Cavaliers team

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='Age', y='Salary', data=df)
plt.title('Correlation between Age and Salary')
plt.show()
```

 There is weak/strong correlation between age and salary, suggesting age may not be the primary driver of high pay compared to specific roles.
